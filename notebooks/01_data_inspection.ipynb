```python
def missing_value_summary(df):
    summary = pd.DataFrame({
        "missing_count": df.isna().sum(),
        "missing_percentage": (
            df.isna().mean() * 100
        ).round(2)
    })

    return summary[summary["missing_count"] > 0].sort_values(
        "missing_percentage",
        ascending=False
    )

print("Netflix missing values:")
display(missing_value_summary(netflix_df))

print("Disney+ missing values:")
display(missing_value_summary(disney_df))
```

## Initial Data Inspection and Cleaning Observations

The Netflix dataset contains 8,807 records and 12 columns, while the Disney+ dataset contains 1,450 records and 12 columns. Both datasets use the same columns in the same order, which will allow them to be cleaned consistently and combined for comparison.

### Data structure

- Both datasets contain the following fields: `show_id`, `type`, `title`, `director`, `cast`, `country`, `date_added`, `release_year`, `rating`, `duration`, `listed_in`, and `description`.
- `release_year` is stored as an integer.
- All other columns are currently stored as text.
- `date_added` must be converted from text to a datetime data type.
- `duration` contains a number and a unit, such as `90 min`, `1 Season`, or `2 Seasons`, and should be separated into numeric and categorical fields.

### Missing values

The Netflix dataset contains missing values in:

- `director`: 2,634 records (29.91%)
- `country`: 831 records (9.44%)
- `cast`: 825 records (9.37%)
- `date_added`: 10 records (0.11%)
- `rating`: 4 records (0.05%)
- `duration`: 3 records (0.03%)

The Disney+ dataset contains missing values in:

- `director`: 473 records (32.62%)
- `country`: 219 records (15.10%)
- `cast`: 190 records (13.10%)
- `date_added`: 3 records (0.21%)
- `rating`: 3 records (0.21%)

The missing director, cast, and country values should not be statistically imputed because these are descriptive categorical fields. They can either remain missing or be labeled `Unknown` when needed for analysis. Missing dates should remain unavailable rather than being estimated.

### Duplicate records

- Neither dataset contains fully duplicated rows.
- Neither dataset contains duplicated `show_id` values within its own platform.
- The same `show_id` values appear across platforms, such as `s1`, so `show_id` will not remain globally unique after the datasets are combined.
- A new platform-specific identifier should be created before combining the datasets.
- Titles appearing on both platforms should not automatically be treated as duplicates because the same title may legitimately appear in both catalogs.

### Categorical inconsistencies

- Netflix contains 6,131 movies and 2,676 TV shows.
- Disney+ contains 1,052 movies and 398 TV shows.
- Three Netflix records contain `74 min`, `84 min`, and `66 min` in the `rating` column. These appear to be misplaced duration values and should be moved to the `duration` column.
- Netflix has four missing ratings before correcting the misplaced values.
- Disney+ has three missing ratings.
- Rating labels should be reviewed and standardized before comparison.
- Text fields may contain leading or trailing whitespace that should be removed.

### Multi-value fields

- The `country` column can contain multiple countries in one record.
- The `listed_in` column can contain multiple genres or categories in one record.
- The `cast` and `director` columns can also contain multiple names.
- These fields should not be separated in the main title-level dataset because doing so would duplicate titles.
- Separate exploded tables may be created later for country-level and genre-level analysis.

### Planned cleaning actions

1. Preserve the original raw datasets without modification.
2. Create working copies for cleaning.
3. Remove unnecessary whitespace from column names and text values.
4. Correct the misplaced Netflix duration values.
5. Convert `date_added` to a datetime data type.
6. Separate `duration` into `duration_value` and `duration_unit`.
7. Add a `platform` column to each dataset.
8. Create a platform-specific content identifier.
9. Handle missing categorical values consistently.
10. Combine the cleaned datasets.
11. Validate row counts, identifiers, data types, and remaining missing values.
12. Export the cleaned title-level dataset to the processed-data folder.
